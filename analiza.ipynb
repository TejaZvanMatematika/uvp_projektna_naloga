```python
import pandas as pd
import os

from glavni_sez import shrani_dat_v_sez
from main import SEZ
```

```python
pd.options.display.max_rows = 20


mapa_csv = 'CSV_datoteke'
mapa_pomozne = 'pomozne_datoteke'
sez = []

if SEZ == []:
    pot = os.path.join(mapa_pomozne, 'sez.text')
    SEZ = shrani_dat_v_sez(pot)

sez_csv = []
for _, _, pot_csv, _ in SEZ:
    preberi = pd.read_csv(pot_csv)
    sez_csv.append(preberi)

men = sez_csv[0]
wemen = sez_csv[1]
men_rapid = sez_csv[2]
wemen_rapid = sez_csv[3]
men_blitz = sez_csv[4]
wemen_blitz = sez_csv[5]

sez_imen = ['men', 'wemen', 'men_rapid', 'wemen_rapid', 'men_blitz', 'wemen_blitz']
sez_podatkov = [men, wemen, men_rapid, wemen_rapid, men_blitz, wemen_blitz]
```

Analizirala bom podatke iz spletne strani FIDE. To je spletna stran ki objavlja ELO rating najbolših igralcev, rezultate šahovskih tekem in novice o šahu. Jaz se bom osredotočila na ELO ratinge igralcev.

```python
sez_moški = []
sez_ženske = []

for i in range(0, len(sez_imen), 2):
    povp = sez_podatkov[i].elo.mean().round()
    sez_moški.append(int(povp))

for i in range(1, len(sez_imen), 2):
    povp = sez_podatkov[i].elo.mean().round()
    sez_ženske.append(int(povp))

tabela = {
    'Kategorija': ['Standard', 'Rapid', 'Blitz'],
    'Moški': sez_moški,
    'Ženske': sez_ženske
}

tabela = pd.DataFrame(tabela)
tabela
```

Opazimo lahko da imajo v vseh kategorijah moški povprečno večji ELO rating kot pa ženske (razlika je za približno 200-300 ELO-ta). Ta velika razlika ni zato ker so ženske manj pametne od moških, ampak zato ker na sploh manj žensk igra šah.
<br>
Ne poznamo točnega razloga za to, ampak največ ljudi meni da je zaradi negativnih stereotipov. Nekateri moški (tudi te ki so na vrhu lestvice) menijo, da šah ni za ženske, da so slabše zaradi spola. Precej šahistov si je izmislilo razlog, zakaj jih je šahistka premagala v igri (npr. nekdo je rekel da bi zmagal, če ga ne bi glava bolela).
